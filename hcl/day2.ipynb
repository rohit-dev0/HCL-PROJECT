```python
#ARGUMENTS
def add(*num):   # single stars used
    sum = 0
    for n in num :
        sum+=n

    return sum     

print(add())         #0 argument
print(add(1))          #1 argument
print(add(1,2,))        #2 argument 
print(add(1,2,3,))        #3 argument
print(add(1,2,3,4))        #4 argument


# KEYWORD ARGUMENTS
```

```python

def display_emp(**data):
    for k, v in data.items():
        print("{} \t : {}".format(k, v))


#calling **kwargs function
display_emp(id=1001, name="hillary" ,age=30) 
```
