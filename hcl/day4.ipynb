```python
# BASIC MATHEMATICAL FUNCTION
# LIKE np.add(), np.subtract(), np.multiply(), np.divide()

import numpy as np
a = np.array([1,2,3])
b = np.array([4,5,6])


print(np.add(a,b))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))
print(np.sum(a))
print(np.min(a))
print(np.max(a))
```
