```python
def square_cube(n):
    square = n*n
    cube = square*n
    return square, cube

s,c = square_cube(5)
print("square of 5 = ",s)
print("cube of 5 = ",c)
```

```python
ans = square_cube(4)
s = ans[0]
c = ans[1]               #using index value can be retrieved

print("square of 4 =",s,"and cube of 4 =",c)
```

```python
def is_valid_patient_id(patient_id_str):
    min_length = 5
    max_length = 10

    if len(patient_id_str)>= min_length and len(patient_id_str) <= max_length:
        return True
    else:
        return False

#SAMPLE PATIENT IDs
patient1 = "P-101"
patient2 = "1234"
patient3 = "P-101-ABCDEF"

print(f"'{patient1}' is valid: {is_valid_patient_id(patient1)}")
print(f"'{patient2}' is valid: {is_valid_patient_id(patient2)}")
print(f"'{patient3}' is valid: {is_valid_patient_id(patient3)}")
```

```python
#PASSWORD CHECKER

def pass_check(pw):
    if len(pw)>=8:
        if pw.isalnum():
            print('valid password')
        else:
            print('you need alphabets and number both')
    else:
        print('length must be >=8 characters ')


pw = 'yghghjbj52626guy121gjh'
pass_check(pw)
```

```python
#OBJECT ORIENTED PROGRAM   CLASS

class Car:
  def __init__(self, name, top_speed, nums_of_doors):
      self.name = name 
      self.top_speed = top_speed
      self.nums_of_doors = nums_of_doors
  def acceleration(self):
      print('speeding up...')
  def details(self):
      print('car name is ', self.name)
      print('top speed is ', self.top_speed)
      print('nums of doors ', self.nums_of_doors)

#creating objects

toyota = Car('Toyota Corolla', 200, 4)
bmw = Car('X6', 280, 4)
swift = Car('Desire', 130, 4)
wagonR = Car('LXI', 100, 4)
rr = Car('Defender', 360, 5)


bmw.details()
toyota.acceleration()
rr.details()

```

```python
#PYTHON INHERITENCE

# SINGLE INHERITENCE

class A: 
  def hello(self):
    print('hello')

class B(A):
  def hi(self):
    print('hi')

obj = B()
obj.hello()
obj.hi()
```

```python
#MULTIPLE INHERITENCE

class A:       #parent clas
  def hello(self):
    print('hello')

class B:     #parent class
  def hi(self):
    print('hi')

class C(A,B):   #child class
  def bye(self):
    print('bye')

 
obj = C()
obj.hello()
obj.hi()
obj.bye()
```
